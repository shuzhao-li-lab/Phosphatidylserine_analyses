### Merge the table & allow for plottings
- x as name
- y as rtime
- `cSelectivity` as size
- `color` as annotation_evidence
- `shape` to label those with ms2 and have neutral loss shown

```python
keeps = ["mz", "rtime", "rtime_left_base", "rtime_right_base", "cSelectivity", "goodness_fitting", "snr","annotation_evidence",'name']

mdf = PS_std_df.loc[:,keeps].transpose().merge(PS_species_annot_df.loc[:,keeps].transpose(),left_index = True,right_index = True)

mdf_t= mdf.transpose()

mdf_t = mdf_t.sort_values("mz")
```
