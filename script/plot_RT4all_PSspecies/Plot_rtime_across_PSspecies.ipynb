```r
library(dplyr)
library(ggplot2)
```

```r
path2table = "./data/cleaned_data.csv"
```

```r
mdf = read.csv(path2table,row.names = 1, check.names = FALSE)
```

```r
mdf %>% head(2)
```

```r
mdf = mdf %>% filter(grepl("MS2|standard", annotation_evidence))
```

```r
# Sorting by 'rtime'
mdf <- mdf %>% arrange(rtime)
```

```r
# Plotting with ggplot2
p <- ggplot(mdf, aes(x = rtime, y = name)) +

  # add error bar
  geom_errorbarh(aes(xmin = rtime_left_base, xmax = rtime_right_base), color = "gray", alpha = 0.5) +
  
  geom_point(aes(size = goodness_fitting, color = annotation_evidence)) +
  
  geom_point(data = mdf %>% filter(grepl("MS2", annotation_evidence)),
             aes(size = cSelectivity), color = "red", fill = NA) +
  scale_color_viridis_d(option = "D") +  
  labs(title = "Retention Time with goodness_fitting as Size and Annotation Evidence as Color",
       x = "Retention Time (rtime)", y = "Name", color = "Annotation Evidence") +
  theme_minimal() +
  theme(axis.text.y = element_text(angle = 0, hjust = 1)) +
  xlim(250, 350)  # Set retention time range from 0 to 600
```

```r
ggsave("./data/final_result.pdf",p,width= 9,height = 5)
```
